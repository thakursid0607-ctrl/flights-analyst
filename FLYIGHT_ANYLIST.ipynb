```python

query2 = """
SELECT a.aircraft_registration, a.aircraft_model, COUNT(f.flight_number) AS total_flights
FROM aircrafts a 
JOIN flights f ON a.aircraft_registration = f.aircraft_registration
GROUP BY a.aircraft_registration, a.aircraft_model
HAVING COUNT(f.flight_number) > 5
ORDER BY total_flights DESC;

"""

cursor_mysql.execute(query2)
result2 = cursor_mysql.fetchall()
print( result2)
```

```python
query3 = """
SELECT a.name, COUNT(f.flight_number) AS outbound_flights
FROM airports a
JOIN flights f ON a.iata_code = f.origin_iata
GROUP BY a.iata_code, a.name
HAVING COUNT(f.flight_number) > 5
ORDER BY outbound_flights DESC;

"""

cursor_mysql.execute(query3)
result3 = cursor_mysql.fetchall()
print( result3)
```

query4 --> Find the top 3 destination airports (name, city) by number of arriving flights, sorted by count descending.


```python
query4 = """
SELECT a.name, a.city, COUNT(f.flight_number) AS arriving_flights
FROM airports a 
JOIN flights f 
    ON a.iata_code = f.destination_iata
GROUP BY a.iata_code, a.name, a.city
ORDER BY arriving_flights DESC
LIMIT 3;

"""

cursor_mysql.execute(query4)
result4 = cursor_mysql.fetchall()
print( result4)



```

query5 --> Show for each flight: number, origin, destination, and a label 'Domestic' or 'International' using CASE WHEN on country match.


```python
query5 = """
SELECT f.flight_number, f.origin_iata, f.destination_iata,
    CASE 
        WHEN a1.country = a2.country THEN 'Domestic'
        ELSE 'International'
    END AS flight_type
FROM flights f
JOIN airports a1 ON f.origin_iata = a1.iata_code
JOIN airports a2 ON f.destination_iata = a2.iata_code;
"""

cursor_mysql.execute(query5)
result5 = cursor_mysql.fetchall()
result5


```

```python
query6 = """
SELECT f.flight_number, f.aircraft_registration AS aircraft,a.name AS departure_airport_name,f.actual_arrival AS arrival_time
FROM flights f
JOIN airports a 
    ON f.origin_iata = a.iata_code                -- Joins to map departure code to airport name
WHERE 
    f.destination_iata = 'DEL'                     -- Filters for flights arriving at Delhi
    AND f.actual_arrival IS NOT NULL               -- Excludes flights that haven't landed yet
ORDER BY 
    f.actual_arrival DESC                          -- Sorts by latest arrival time first
LIMIT 5;                                           -- Restricts output to the top 5 records
"""

cursor_mysql.execute(query6)
result6 = cursor_mysql.fetchall()
result6


```

query11 -->  For each destination airport, compute the % of delayed flights (status='Delayed') among all arrivals, sorted by highest percentage. 

```python
import mysql.connector

# 1. Open the connection
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="Sid@9358805903"
)

# 2. Open a standard cursor (Fixed: removed trailing comma and buffered arg)
cursor = conn.cursor()

# 3. Explicitly select your database schema
cursor.execute("USE flights_analyst;")


query11 = """
SELECT 
    f.destination_iata AS destination_airport,
    COUNT(CASE WHEN f.scheduled_departure < f.actual_departure THEN 1 END) AS delayed_arrivals,
    COUNT(f.flight_number) AS total_arrivals,
    ROUND(
        (COUNT(CASE WHEN f.scheduled_departure < f.actual_departure THEN 1 END) * 100.0) / COUNT(f.flight_number), 
        2
    ) AS delay_percentage
FROM 
    flights f
GROUP BY 
    f.destination_iata
ORDER BY 
    delay_percentage DESC
"""


try:
    # 5. Execute and instantly fetch to clear the buffer
    cursor.execute(query11)
    result11 = cursor.fetchall()
    
    
    for row in result11[:5]:
        print(row)
        
finally:
    # 6. Always close the cursor and connection right away
    cursor.close()
    conn.close()
```
